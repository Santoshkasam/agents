# 2. Build a simple chat interface

```python
system_prompt = f"""

# Your role
You are a digital twin of the person you are representing. You live in a website, and answers questions of the website visitors.
The website belongs to the person you are representing. 
You answer questions related to their career, background, skills and experience.

Here are the details of the person you are representing

{summary}

# Context

Here is the summary of the person's detailed resume.

{resume}

# Rules

Engage with the user professionally and engagingly, as if you are speaking to a potential client or a future employer.
Avoid answering questions that are not related to the user's career, background, skills and experience. Steer the conversation back to the professional topics.

Always stay in charecter as the digital twin of the person you are representing.

Important: If you don't have an answer, say so. Do not make up the answer.
If you can't find the answer in the context, say that you don't know.
"""

display(Markdown(system_prompt))
```

```python
messages = [
    {"role":"system", "content":system_prompt},
    {"role":"user",   "content": "Hi, introduce yourself briefly"}
]

response = openai.chat.completions.create(model="gpt-5.4-nano", messages = messages)
display(Markdown(response.choices[0].message.content))
```

```python
def chat(message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model="gpt-4.1-mini", messages = messages)
    return response.choices[0].message.content
```

```python
chat("please summarize who you are", [])
```

```python
# Create a chat interface UI using gradio

gr.ChatInterface(chat).launch(inbrowser=True)
```

# 3. Add Tools..

```python
# Define a function that acts as a tool to record email adress in a txt file

def record_email_tool(email):
    print(f"Tool called to record an email: {email}")
    with open("emails.txt", "a", encoding="utf-8") as f:
        f.write(email + "\n")
    return "Email recieved"
```

```python
record_email_tool("nikithags19@gmail.com")
```

## Step.1 - Write a json to describe the tool 

```python
record_email_tool_json = {
    "name": "record_email_tool",
    "description": "Use this tool to record that a user provided their email address",
    "parameters": {
        "type": "object",
        "properties": {
            "email": {"type": "string", "description": "The email address of this user"}
        },
        "required":["email"],
        "additionalProperties": False
    }
}
```

```python
# Define the tools list that will be later passed to an llm
tools = [{"type": "function", "function": record_email_tool_json}]
tools
```

## Step 2 - A new chat() function

- This is where we implement tool call. Tool call is basically a if statement. 

- The tool is actually called by the function we define (or by the function we use from agent sdk frameworks), based on the llm response. The llm cannot call the tool directly by itself. 

```python
def chat(message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model="gpt-5.4-mini", messages = messages, tools=tools)

    # Check for tool call by llm
    if response.choices[0].finish_reason=="tool_calls":
        # print("LLM made a tool_calls response")

        message = response.choices[0].message
        # print(f"Message: {message}")
        
        tool_call = message.tool_calls[0]
        # print(f"Tool call - 0: {tool_call}")

        email = json.loads(tool_call.function.arguments).get("email")
        tool_response = record_email_tool(email)
        messages.append(message)
        messages.append({"role": "tool", "content": tool_response, "tool_call_id": tool_call.id})
        response = openai.chat.completions.create(model="gpt-5.4-mini", messages = messages, tools=tools)
    return response.choices[0].message.content
```

```python
gr.ChatInterface(chat).launch(inbrowser=True)
```
